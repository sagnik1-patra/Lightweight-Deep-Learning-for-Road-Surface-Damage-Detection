```python
# ==============================================================
# CELL 9: COUNT POTHOLE BOUNDING BOXES
# ==============================================================

def count_boxes(label_folder):

    total_boxes = 0

    for label_file in label_folder.glob("*.txt"):

        with open(label_file, "r") as file:
            lines = [
                line.strip()
                for line in file.readlines()
                if line.strip()
            ]

        total_boxes += len(lines)

    return total_boxes


train_boxes = count_boxes(TRAIN_LABELS)
val_boxes = count_boxes(VAL_LABELS)
test_boxes = count_boxes(TEST_LABELS)

total_boxes = train_boxes + val_boxes + test_boxes


print("Train potholes :", train_boxes)
print("Valid potholes :", val_boxes)
print("Test potholes  :", test_boxes)
print("Total potholes :", total_boxes)
```

```python
# ==============================================================
# CELL 12: ANALYZE BOUNDING BOX SIZES
# ==============================================================

bbox_records = []


def analyze_labels(label_folder, split):

    for label_file in label_folder.glob("*.txt"):

        with open(label_file, "r") as file:

            for line in file:

                values = line.strip().split()

                if len(values) != 5:
                    continue

                class_id, x, y, width, height = map(
                    float,
                    values
                )

                bbox_records.append({
                    "split": split,
                    "file": label_file.name,
                    "class_id": int(class_id),
                    "x_center": x,
                    "y_center": y,
                    "width": width,
                    "height": height,
                    "area": width * height
                })


analyze_labels(TRAIN_LABELS, "train")
analyze_labels(VAL_LABELS, "validation")
analyze_labels(TEST_LABELS, "test")


bbox_analysis_df = pd.DataFrame(bbox_records)

bbox_analysis_df.to_csv(
    CSV_DIR / "bounding_box_analysis.csv",
    index=False
)

bbox_analysis_df.head()
```

```python
# ==============================================================
# CELL 14: DISPLAY SAMPLE TRAINING IMAGES
# ==============================================================

def draw_ground_truth(image_path, label_path):

    image = cv2.imread(str(image_path))
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    height, width = image.shape[:2]

    if label_path.exists():

        with open(label_path, "r") as file:

            for line in file:

                values = line.strip().split()

                if len(values) != 5:
                    continue

                class_id, xc, yc, bw, bh = map(
                    float,
                    values
                )

                x1 = int((xc - bw / 2) * width)
                y1 = int((yc - bh / 2) * height)
                x2 = int((xc + bw / 2) * width)
                y2 = int((yc + bh / 2) * height)

                cv2.rectangle(
                    image,
                    (x1, y1),
                    (x2, y2),
                    (255, 0, 0),
                    2
                )

                cv2.putText(
                    image,
                    "Pothole",
                    (x1, max(20, y1 - 5)),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.6,
                    (255, 0, 0),
                    2
                )

    return image
```

```python
# ==============================================================
# CELL 35: BUILD DETECTION-LEVEL PREDICTION CSV
# ==============================================================

prediction_records = []


for result in tqdm(
    prediction_results,
    desc="Processing predictions"
):

    image_name = Path(
        result.path
    ).name

    image_height, image_width = (
        result.orig_shape
    )


    if result.boxes is None:
        continue


    boxes = result.boxes


    for detection_id in range(
        len(boxes)
    ):

        xyxy = (
            boxes.xyxy[detection_id]
            .cpu()
            .numpy()
        )

        confidence = float(
            boxes.conf[detection_id]
            .cpu()
            .numpy()
        )

        class_id = int(
            boxes.cls[detection_id]
            .cpu()
            .numpy()
        )


        x1, y1, x2, y2 = xyxy


        prediction_records.append({

            "image": image_name,

            "detection_id":
                detection_id + 1,

            "class_id":
                class_id,

            "class_name":
                best_model.names[
                    class_id
                ],

            "confidence":
                confidence,

            "x1":
                float(x1),

            "y1":
                float(y1),

            "x2":
                float(x2),

            "y2":
                float(y2),

            "box_width":
                float(x2 - x1),

            "box_height":
                float(y2 - y1),

            "box_area":
                float(
                    (x2 - x1) *
                    (y2 - y1)
                ),

            "image_width":
                image_width,

            "image_height":
                image_height
        })


prediction_df = pd.DataFrame(
    prediction_records
)


prediction_df.to_csv(
    CSV_DIR / "detection_predictions.csv",
    index=False
)


print(
    "Total potholes detected:",
    len(prediction_df)
)

display(
    prediction_df.head(20)
)
```

```python
# ==============================================================
# CELL 36: CREATE IMAGE LEVEL PREDICTION RESULTS
# ==============================================================

image_prediction_records = []


for result in prediction_results:

    image_name = Path(
        result.path
    ).name


    if (
        result.boxes is not None
        and len(result.boxes) > 0
    ):

        confidences = (
            result.boxes.conf
            .cpu()
            .numpy()
        )

        detection_count = len(
            confidences
        )

        max_confidence = float(
            confidences.max()
        )

        mean_confidence = float(
            confidences.mean()
        )

        detected = 1

    else:

        detection_count = 0
        max_confidence = 0
        mean_confidence = 0
        detected = 0


    image_prediction_records.append({

        "image":
            image_name,

        "pothole_detected":
            detected,

        "number_of_potholes":
            detection_count,

        "maximum_confidence":
            max_confidence,

        "mean_confidence":
            mean_confidence
    })


image_predictions_df = pd.DataFrame(
    image_prediction_records
)


image_predictions_df.to_csv(
    CSV_DIR / "image_level_predictions.csv",
    index=False
)


display(
    image_predictions_df.head(20)
)
```

```python
# ==============================================================
# CELL 49: CREATE FINAL RESULTS TABLE - FIXED VERSION
# ==============================================================

import numpy as np
import pandas as pd
import json
import yaml
from pathlib import Path


# --------------------------------------------------------------
# Function to convert NumPy / Path objects into standard Python
# objects so that JSON and YAML can save them safely.
# --------------------------------------------------------------

def convert_to_python_type(obj):

    # NumPy integer -> Python int
    if isinstance(obj, np.integer):
        return int(obj)

    # NumPy float -> Python float
    elif isinstance(obj, np.floating):
        return float(obj)

    # NumPy boolean -> Python bool
    elif isinstance(obj, np.bool_):
        return bool(obj)

    # NumPy array -> Python list
    elif isinstance(obj, np.ndarray):
        return obj.tolist()

    # pathlib.Path -> string
    elif isinstance(obj, Path):
        return str(obj)

    # Dictionary -> recursively convert values
    elif isinstance(obj, dict):
        return {
            str(key): convert_to_python_type(value)
            for key, value in obj.items()
        }

    # List -> recursively convert items
    elif isinstance(obj, list):
        return [
            convert_to_python_type(value)
            for value in obj
        ]

    # Tuple -> convert into list
    elif isinstance(obj, tuple):
        return [
            convert_to_python_type(value)
            for value in obj
        ]

    # Normal Python object
    else:
        return obj


# ==============================================================
# CREATE FINAL RESULTS
# ==============================================================

final_results = {

    "Model":
        "YOLOv8n",

    "Task":
        "Pothole Detection",

    "Research_Title":
        "Lightweight Deep Learning for Road Surface Damage Detection",

    "Train_Images":
        int(train_count),

    "Validation_Images":
        int(val_count),

    "Test_Images":
        int(test_count),

    "Total_Dataset_Images":
        int(total_count),

    "Total_Annotated_Potholes":
        int(total_boxes),

    "Training_Time_Minutes":
        float(training_time / 60),

    # --------------------------
    # Validation Results
    # --------------------------

    "Validation_Precision":
        float(validation_results["precision"]),

    "Validation_Recall":
        float(validation_results["recall"]),

    "Validation_mAP50":
        float(validation_results["mAP50"]),

    "Validation_mAP50_95":
        float(validation_results["mAP50_95"]),

    # --------------------------
    # Test Results
    # --------------------------

    "Test_Precision":
        float(test_results["precision"]),

    "Test_Recall":
        float(test_results["recall"]),

    "Test_mAP50":
        float(test_results["mAP50"]),

    "Test_mAP50_95":
        float(test_results["mAP50_95"]),

    # --------------------------
    # Lightweight Model Results
    # --------------------------

    "Model_Size_MB":
        float(model_size_mb),

    "Average_Inference_Time_ms":
        float(average_inference_ms),

    "Approximate_FPS":
        float(fps),

    # --------------------------
    # Prediction Results
    # --------------------------

    "Mean_Absolute_Count_Error":
        float(mean_absolute_count_error),

    "Number_of_Test_Detections":
        int(len(prediction_df))
}


# Convert everything again for complete safety
final_results = convert_to_python_type(
    final_results
)


# ==============================================================
# CREATE DATAFRAME
# ==============================================================

final_results_df = pd.DataFrame(
    [final_results]
)


# ==============================================================
# SAVE FINAL RESULTS CSV
# ==============================================================

final_csv_path = (
    CSV_DIR /
    "final_research_results.csv"
)


final_results_df.to_csv(
    final_csv_path,
    index=False
)


print(
    "Final results CSV saved:"
)

print(
    final_csv_path
)


print("\nFINAL RESEARCH RESULTS")
print("=" * 70)


display(
    final_results_df.T
)
```
